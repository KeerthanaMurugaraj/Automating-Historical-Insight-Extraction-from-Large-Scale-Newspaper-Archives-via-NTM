# Top2Vec: fully preprocessed text (word-level)

```python
%cd ../..
```

```python
word_embeddings_mpnet=np.load("../../datasets/compare_study_dataset/1987_2002/aggregation/word_embeddings_octis_MPnet.npy")
word_embeddings_distilroberta=np.load("../../datasets/compare_study_dataset/1987_2002/aggregation/word_embeddings_octis_distilroberta.npy")
word_embeddings_alibaba=np.load("../../datasets/compare_study_dataset//1987_2002/aggregation/word_embeddings_octis_Alibaba-NLP_gte-base-en-v1.5.npy")
```

```python
# MPNET
import pickle
from sentence_transformers import SentenceTransformer

for i in range(1):
    dataset, custom = "../../datasets/compare_study_dataset/1987_2002/aggregation/word_embedding_data_384", True
    model = SentenceTransformer("all-mpnet-base-v2")
    print(model.max_seq_length)
    chunk_length=model.max_seq_length
    params = {"nr_topics": [(i+1)*10 for i in range(5)],"embedding_model":"all-mpnet-base-v2",
              "split_documents":'True',"document_chunker":'sequential',"use_embedding_model_tokenizer":'True',
              "index_topics":'True',"chunk_length":chunk_length,
              # "embedding_model": "all-mpnet-base-v2",
              "hdbscan_args": {'min_cluster_size': 15,
                               'metric': 'euclidean',
                               'cluster_selection_method': 'eom'}}
    start_time = datetime.now()
    print("Start_time",start_time)
    params_results = []
    trainer = Trainer(dataset=dataset,
                      custom_dataset=custom,
                      custom_model=Top2Vec_custom,
                      bt_embeddings=word_embeddings_mpnet,
                      model_name="Top2Vec",
                      params=params,
                      verbose=True)
    results, output_mpnet = trainer.train(save=f"../evaluation_results/Results_Comp_Study/1987_2002/Top2Vec/word/Top2Vec_mpnet_{i+1}")
    params_results.append((params, results))
    print("Score:", results[0]["Scores"])
    end_time = datetime.now()
    print("End_time",end_time)

    
    execution_time = end_time - start_time
    print(f" Top2Vec-all-mpnet-base-v2 Model execution time _ 50 topics: {execution_time}")

    with open (f'../evaluation_results/Results_Comp_Study/1987_2002/Top2Vec/word/Top2Vec_all-mpnet-base-v2_default_50_TOPICS_{i+1}.pkl', 'wb') as file:
        pickle.dump(output_mpnet,file)
    
    with open('../evaluation_results/Results_Comp_Study/1987_2002/Top2Vec/word/time/Top2Vec-all-mpnet-base-v2_execution_time.txt', 'w') as file:
        file.write(f"Start_time: {start_time} \n end time: {end_time}  Total execution time: {execution_time}\n")
        
    with open('../evaluation_results/Results_Comp_Study/1987_2002/Top2Vec/word/Top2Vec-default-all-mpnet-base-v2_params_results.txt', 'w') as file:
        file.write(f"results: { params_results} , Scores: {results}\n")
```
